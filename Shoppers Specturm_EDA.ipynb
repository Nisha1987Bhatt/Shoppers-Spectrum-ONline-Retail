# **Project Summary -**

The "Shopper Spectrum" project is a comprehensive Exploratory Data Analysis (EDA) focused on an online retail dataset. In the highly competitive e-commerce landscape, understanding customer behavior, purchasing patterns, and sales dynamics is critical for sustainable growth and targeted marketing. The primary objective of this project is to dive deep into transactional data to uncover hidden patterns, segment shoppers based on their purchasing habits, and provide actionable, data-driven insights that can help optimize inventory management, improve customer retention, and maximize overall revenue.
Data Cleaning and Preprocessing
Before diving into analysis, a rigorous data cleaning pipeline was implemented to ensure data integrity and quality. The raw transactional dataset presented several real-world anomalies, such as missing values (specifically in critical identifiers like CustomerID), duplicate records, and negative values in quantities and unit prices—which typically represent order cancellations, returns, or administrative corrections.
Exploratory Data Analysis (EDA) Framework
Following a structured UBM (Univariate, Bivariate, and Multivariate) analysis approach, more than 15 logical and meaningful visualizations were constructed to map the entire "Shopper Spectrum."

1. Univariate Analysis
Initial explorations focused on understanding the distribution of individual variables. This included analyzing the frequency distributions of order quantities, unit prices, and transactions across different time periods. It revealed heavily skewed distributions, highlighting that a small percentage of high-volume transactions and popular products drive a disproportionate share of total revenue.

2. Bivariate Analysis
Bivariate relationships were explored to connect time-based and categorical factors with sales metrics.

Temporal Trends: Monthly and daily sales trends were plotted to identify peak shopping seasons (such as holiday rushes) and optimal days of the week for promotional campaigns.

Geographical Insights: Comparing sales performance across different countries clearly distinguished primary high-value domestic markets from emerging international clusters.

3. Multivariate Analysis
Advanced combinations of variables were analyzed to unlock deeper behavioral layers. By mapping transaction value against both time-of-day and customer segments, we identified when high-value wholesale buyers are most active compared to retail consumers.

Key Insights & Business Impact
The data revealed several critical business trends:

The Pareto Principle in Action: A small, elite segment of customers contributes to the majority of total sales value. Implementing an RFM (Recency, Frequency, Monetary) framework allows the business to isolate these high-value segments for exclusive loyalty programs.

Return Rate Correlation: High return rates were observed in specific product categories. Identifying these clusters allows quality control teams or suppliers to rectify product descriptions or sizing issues, turning a revenue leak into a cost-saving opportunity.

Peak Ordering Windows: Transaction volume peaks during midday hours (10:00 AM to 2:00 PM). Marketing teams can capitalize on this by scheduling email newsletters and flash sales right before these high-activity windows to boost conversion rates.

## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Write your code to make your dataset analysis ready.
import pandas as pd
import numpy as np
#load dataset
df = pd.read_csv(r"C:\Users\Ashutosh Bhatt\OneDrive - Qualicentric ITES Private Limited\Desktop\Shopper specturem_ecommerce\online_retail.csv")
# check for missing values
print (f"Mising Values : {df.isnull().sum()}")
# repalce all -1 sentinels
print(f" Replace all Sentinels : {df.replace(1,np.nan,inplace=True)}")
print(f" Replace all -1 Sentinels :{df.replace("-1",np.nan,inplace=True)}")

# extracting time from Invoice date column making a new column time and day
# convert InvoiceDate to datetime format
df['InvoiceDate'] = pd.to_datetime(df['InvoiceDate'])
#Extract the time H:M:S
df['Time'] = df['InvoiceDate'].dt.time
#Extract the day 
df['Day'] = df['InvoiceDate'].dt.day_name()
# removing time from the InvoiceDate column
df['InvoiceDate'] = df['InvoiceDate'].dt.date
# calculate the total_price
df['Total_Price'] =df['Quantity']*df['UnitPrice']
# drop rows where columns have missing values
df_clean = df.dropna(subset=['Total_Price','Country',])
#Revenve Contribution %
df['Revenve_Contribution'] = (df['Total_Price']/df['Total_Price'].sum())*100
# creating a flag for cancelled orders
df['Is_Cancelled'] = df['InvoiceNo'].astype(str).str.startswith('C')
# checking  how many cancellation
print(f"Number of Cancellation :{df['Is_Cancelled'].sum()}")
# cleaning the whitespaces in string columns
df['Description'] = df['Description'].str.strip()
df['StockCode'] = df['StockCode'].str.strip()
# removing duplicates
print(f"Removing Dupicates : {df.drop_duplicates(inplace=True)}")
# check the data
df.head()
# saving the cleaned data
df.to_csv("Shopper Spectrum_cleaned_data.csv",index=False)
print("Saved the Dataset")
````

````python
# Chart - 2 visualization code
# line plot showing relation between quantity and total price 
plt.figure(figsize=(10,6))
plt.plot("Quantity","Total_Price",data=df ,alpha=0.5,color="red")
plt.xlabel('Quantity')
plt.ylabel('Total Price')
plt.title('2.Relation between Quantity and Price-Retail View')
plt.show()
````

##### 1. Why did you pick the specific chart?

Answer Here.* **Ideal for Continuous Numerical Data:** A line Plot is the mathematically correct chart type for analyzing the underlying relationship, distribution, or correlation between two continuous numeric variables (`Quantity` and `UnitPrice`). 
* **Overlapping Identification via Density:** By using an `alpha` parameter (transparency), individual data points blend together. Highly dense regions turn into dark, saturated clusters, visually revealing where the vast majority of everyday customer transactions occur.
* **Outlier Isolation:** Zooming into specific axes limits (Quantities < 100, Unit Prices < £50) prevents the plot from being compressed by extreme wholesale bulk orders and deep negative returns, creating a clean retail view.


##### 2. What is/are the insight(s) found from the chart?

Answer Here-* **The Inverse Pricing Law:** A clear boundary emerges showing that high-priced items (above £15–£20) are strictly purchased in small single-digit quantities (1 to 5 units). Mass volume purchases (quantities over 30) only trigger when the unit price drops significantly below £5.
* **The Core Retail "Sweet Spot":** The heaviest concentration of transactions is tightly locked within a zone where the Unit Price is under £10 and the Quantity is under 20 units per line item. This represents the store's baseline everyday shopper profile.
* **Structured Case-Size Purchasing:** The data points form clear, vertical columns at quantities like 12, 24, 36, and 48. This explicitly proves that bulk buyers or standard consumers frequently purchase items in structured pre-packaged box increments (dozens) rather than completely random counts.

##### 3. Will the gained insights help creating a positive business impact?
Are there any insights that lead to negative growth? Justify with specific reason.

Answer Here-Positive Business Impacts:**
* **Inventory & Tiered Pricing Strategies:** Recognizing the vertical buying patterns at quantities of 12, 24, and 36 allows marketing to introduce "Buy a Case and Save" tiered discounts, directly boosting the Average Order Value (AOV).
* **Smart Recommendations:** Knowing that higher-priced items are only bought in low quantities means the e-commerce recommendation engine should avoid pushing high-volume bundles for expensive items, focusing instead on cross-selling complementary low-cost accessories.

**Risks of Negative Growth (Justification):**
* **The Bulk Stockout Trap:** If logistics and inventory management fail to keep up with the high-volume/low-price products (the ones bought in counts of 30+), the store faces a rapid out-of-stock scenario. Because these items form the bulk of processing transactions, missing stock leads to dropped carts and immediate negative growth.
* **Margin Cannibalization:** Pushing volume discounts too aggressively on the low-price "sweet spot" products can backfire. If users shift entirely to low-margin items, it can cannibalize the sales of higher-priced, more profitable goods, lowering the business's net profit margins

### 2. Handling Outliers

````python
# Handling Outliers & Outlier treatments
# Handling Outliers & Outlier treatments
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
df = pd.read_csv(r"C:\Users\Ashutosh Bhatt\OneDrive - Qualicentric ITES Private Limited\Desktop\Shopper specturem_ecommerce\Shopper Spectrum_imputed _data ML.csv")
print(f" Dataset Shape : {df.shape}")
print("="*55)
numeric_cols=['Quantity','UnitPrice']
# detetcing outliers
fig,axes=plt.subplots(1,len(numeric_cols),figsize=(20,5))
fig.suptitle("Boxplot -outliers detection defore treatment", fontsize=14)
for i , col in enumerate(numeric_cols):
    axes[i].boxplot(df[col].dropna())
    axes[i].set_title(col)
    axes[i].set_xlabel(col)
plt.tight_layout()
plt.show()
print("="*55)
# detecting outliers using IQR method
print("IQR method for outliers detection")
outlier_report=[]
for col in numeric_cols:
    Q1=df[col].quantile(0.25)
    Q3=df[col].quantile(0.75)
    IQR = Q3-Q1

    lower = Q1 - 1.5*IQR
    upper = Q3 + 1.5*IQR

    outlier=df[(df[col]< lower)|(df[col]>upper)]
    outlier_report.append({
        "Column"       : col,
        "Q1"           : round(Q1,2),
        "Q3"           : round(Q3,2),
        "IQR"          : round(IQR,2),
        "Lower Bound"  : round(lower,2),
        "Upper Bound"  : round(upper,2),
        "Outlier Count": len(outlier),
         "Outlier %"   : round(len(outlier) / len(df) * 100, 2)
    })

outlier_df=pd.DataFrame(outlier_report).set_index("Column")
print(outlier_df.to_string())
print("="*55)

# Outlier Treatment method
print("Outlier Treatment Method")
df_treated=df.copy()
# Treatment-1 Capping / Winsorization (Best for salary)
print("Treatment Capping (Winsorization)")
cap_cols=["Quantity","UnitPrice"]
for col in cap_cols:
    Q1= df_treated[col].quantile(0.25)
    Q3= df_treated[col].quantile(0.75)
    IQR= Q3-Q1

    lower=Q1-1.5*IQR
    upper=Q3+1.5*IQR

    df_treated[col]=df_treated[col].clip(lower=lower,upper=upper)
    after = df_treated[col].describe()
    print(f"   {col:<15} capped [{lower:,.0f} to {upper:,.0f}]")
df_treated['Total_Price'] = df_treated['Quantity']*df_treated['UnitPrice']
df_treated['Revenue_Contribution'] = (df_treated['Total_Price'] / df_treated['Total_Price'].sum()) * 100
#Visualize before and after Treatment
print("Before Vs After Treatment")
print("="*55) 

fig,axes = plt.subplots(1,len(cap_cols), figsize=(15,10))
fig.suptitle("After Outlier Treatment", fontsize =14)
for i , col in enumerate(cap_cols):   
    axes[i].boxplot(df_treated[col].dropna())
    axes[i].set_title(f"{col}AFTER")
    axes[i].set_xlabel("After Capping")

print(" Treated Outliers")
df_treated.to_csv("Shoppers Spectrum_treated outliers.csv")
print("Saved")
````

##### What all outlier treatment techniques have you used and why did you use those techniques?

Answer Here.Upon analyzing the numerical features via boxplots and the Interquartile Range (IQR) method, it was revealed that our e-commerce dataset contained a substantial number of extreme values (12.24% of rows in Quantity and 7.35% in UnitPrice).
To address this, i implemented a combination of Capping (Winsorization) and Deterministic Recalculation rather than trimming (dropping) the data.
1. Capping / Winsorization (Via the IQR Method)Columns Applied: Quantity and UnitPrice
Used it:Preservation of Data Volume: Dropping rows with outliers would result in losing nearly 100,000 vital transaction records (over 12% of the dataset). Capping allows us to retain every single transaction while muting the destructive impact of extreme values on down-the-line machine learning algorithms.
Mitigating Extreme Skewness: In e-commerce, a handful of wholesale orders or accidental bulk entries (e.g., millions of items in a single transaction) pull data distributions wildly out of proportion. Using the .clip() function to bound values between the statistical lower limit ($Q1 - 1.5 \times IQR$) and upper limit ($Q3 + 1.5 \times IQR$) keeps values within a realistic operational range.
Handling Skewed Distributions: Unlike Z-score capping, which assumes a perfectly symmetrical normal bell curve, IQR-based capping is completely robust to heavily skewed data distributions typical of transactional quantities and unit prices.
